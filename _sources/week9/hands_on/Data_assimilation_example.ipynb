```python
import numpy as np

# Define a simple nonlinear model
def nonlinear_model(x):
    return np.array([np.sin(x[0]), x[1]**2])

# Tangent Linear Model (Jacobian of the nonlinear model)
def tangent_linear_model(x):
    return np.array([[np.cos(x[0]), 0],
                     [0, 2 * x[1]]])

# Adjoint Model (transpose of the Jacobian)
def adjoint_model(jacobian, gradient):
    return jacobian.T @ gradient

# Cost function: Measure model-observation misfit
def cost_function(x, y_obs, H):
    y_model = H @ nonlinear_model(x)
    return 0.5 * np.linalg.norm(y_model - y_obs)**2

# Gradient of the cost function using adjoint model
def cost_function_gradient(x, y_obs, H):
    y_model = H @ nonlinear_model(x)
    observation_misfit = y_model - y_obs
    jacobian = H @ tangent_linear_model(x)
    gradient = adjoint_model(jacobian, observation_misfit)
    return gradient

# Observation operator
H = np.eye(2)  # Direct observation of both variables

# Observations (synthetic)
y_obs = np.array([0.5, 1.0])

# Initial guess
x0 = np.array([1.0, 0.5])

# Optimization loop
x = x0.copy()
learning_rate = 0.1
tolerance     = 1e-6
max_iter      = 100

print("Initial state:", x)
for iteration in range(max_iter):
    cost = cost_function(x, y_obs, H)
    grad = cost_function_gradient(x, y_obs, H)
    x -= learning_rate * grad  # Update the state

    print(f"Iteration {iteration+1}, Cost: {cost:.6f}, State: {x}")

    if np.linalg.norm(grad) < tolerance:
        print("Converged!")
        break

print("Assimilated state:", x)
```
